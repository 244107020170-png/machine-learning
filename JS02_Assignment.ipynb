```python
import pandas as pd
from sklearn.datasets import load_breast_cancer

# Load dataset dari sklearn
data = load_breast_cancer()

# Buat DataFrame
df = pd.DataFrame(data.data, columns=data.feature_names)

# Tambahkan kolom diagnosis (0 = malignant, 1 = benign)
df['diagnosis'] = data.target

# Cek 5 baris pertama
df.head()
```

```python
print("Jumlah fitur:", len(data.feature_names))
print("Fitur usable:", data.feature_names)
```

```python
# Ubah 0/1 jadi M/B kalau mau lihat
df['diagnosis_label'] = df['diagnosis'].map({0:'Malignant', 1:'Benign'})

# Encode diagnosis jadi 0/1 (M=0, B=1)
df['diagnosis'] = df['diagnosis_label'].map({'Malignant':0, 'Benign':1})

df[['diagnosis_label','diagnosis']].head()
```

```python
from sklearn.preprocessing import StandardScaler

# Pisahkan fitur numerik
features = df.drop(columns=['diagnosis','diagnosis_label'])

# Standardisasi
scaler = StandardScaler()
scaled = scaler.fit_transform(features)

# Buat DataFrame hasil scaling
df_scaled = pd.DataFrame(scaled, columns=features.columns)

# Tambahkan diagnosis
df_scaled['diagnosis'] = df['diagnosis']

df_scaled.head()
```

```python
# Cek statistik setelah standardisasi
df_scaled.describe()

# Pastikan diagnosis tetap ada
df_scaled['diagnosis'].value_counts()
```
