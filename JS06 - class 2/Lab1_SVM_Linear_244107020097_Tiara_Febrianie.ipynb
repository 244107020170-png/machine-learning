### Objective
Understand the fundamental operational principles of Support Vector Machine (SVM) on linearly separable data, the concept of decision boundaries, maximum margin hyperplane, and the role of support vectors.

## Step 1 - Import Libraries

```python
# import libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import seaborn as sns

sns.set_theme(style='whitegrid')
%matplotlib inline
```

## Step 2 - Create Dummy Data
In this practical exercise, we generate a synthetic dataset with two clearly separated classes using `make_blobs`.

```python
from sklearn.datasets import make_blobs
X, y = make_blobs(n_samples=50, centers=2,
                  random_state=0, cluster_std=0.60)
plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn', edgecolors='k')
plt.title('Linearly Separable Synthetic Data')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

## Step 3 - Illustration of Decision Boundary
Multiple linear decision boundaries can separate these two classes. However, different boundaries lead to different classifications for new unseen data points.

```python
xfit = np.linspace(-1, 3.5)
plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn', edgecolors='k')
plt.plot([0.6], [2.1], 'x', color='red', markeredgewidth=2, markersize=10)

for m, b in [(1, 0.65), (0.5, 1.6), (-0.2, 2.9)]:
    plt.plot(xfit, m * xfit + b, '-k')

plt.xlim(-1, 3.5)
plt.title('Multiple Possible Decision Boundaries')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

## Step 4 - Margin Illustration
SVM selects the optimal decision boundary by maximizing the margin (the distance to the nearest training points of any class).

```python
xfit = np.linspace(-1, 3.5)
plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn', edgecolors='k')

for m, b, d in [(1, 0.65, 0.33), (0.5, 1.6, 0.55), (-0.2, 2.9, 0.2)]:
    yfit = m * xfit + b
    plt.plot(xfit, yfit, '-k')
    plt.fill_between(xfit, yfit - d, yfit + d, edgecolor='none',
                     color='#AAAAAA', alpha=0.4)

plt.xlim(-1, 3.5)
plt.title('Margins of Different Decision Boundaries')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

## Step 5 - Model Fitting
Train an SVM model with a linear kernel and a large penalty parameter $C$ (hard margin).

```python
from sklearn.svm import SVC
model = SVC(kernel='linear', C=1E10)
model.fit(X, y)
```

### Defining the Visualization Function

```python
def plot_svc_decision_function(model, ax=None, plot_support=True):
    if ax is None:
        ax = plt.gca()
    xlim = ax.get_xlim()
    ylim = ax.get_ylim()
    
    # create a grid for model evaluation
    x = np.linspace(xlim[0], xlim[1], 30)
    y = np.linspace(ylim[0], ylim[1], 30)
    Y, X_grid = np.meshgrid(y, x)
    xy = np.vstack([X_grid.ravel(), Y.ravel()]).T
    P = model.decision_function(xy).reshape(X_grid.shape)
    
    # plot boundary and margins
    ax.contour(X_grid, Y, P, colors='k',
               levels=[-1, 0, 1], alpha=0.5,
               linestyles=['--', '-', '--'])
    
    # plot support vectors
    if plot_support:
        ax.scatter(model.support_vectors_[:, 0],
                   model.support_vectors_[:, 1],
                   s=300, linewidth=1, edgecolors='k', facecolors='none')
    ax.set_xlim(xlim)
    ax.set_ylim(ylim)
```

### Visualizing Decision Boundary and Margin

```python
plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn', edgecolors='k')
plot_svc_decision_function(model)
plt.title('SVM Decision Boundary with Optimal Margin and Support Vectors')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

### Influence of Sample Size on Decision Boundary
Only support vectors determine the decision boundary. Points far from the margin do not affect the hyperplane.

```python
def plot_svm(N=10, ax=None):
    X_synth, y_synth = make_blobs(n_samples=200, centers=2,
                                  random_state=0, cluster_std=0.60)
    X_sub = X_synth[:N]
    y_sub = y_synth[:N]
    model_sub = SVC(kernel='linear', C=1E10)
    model_sub.fit(X_sub, y_sub)
    
    ax = ax or plt.gca()
    ax.scatter(X_sub[:, 0], X_sub[:, 1], c=y_sub, s=50, cmap='autumn', edgecolors='k')
    ax.set_xlim(-1, 4)
    ax.set_ylim(-1, 6)
    plot_svc_decision_function(model_sub, ax)

fig, ax = plt.subplots(1, 2, figsize=(16, 6))
fig.subplots_adjust(left=0.0625, right=0.95, wspace=0.1)
for axi, N in zip(ax, [60, 120]):
    plot_svm(N, axi)
    axi.set_title(f'N = {N}')
plt.show()
```

### Interactive Exploration with ipywidgets

```python
from ipywidgets import interact, fixed
# interact allows dynamic inspection of sample size variation
interact(plot_svm, N=[10, 200], ax=fixed(None))
plt.show()
```
